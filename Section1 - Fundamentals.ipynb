## Tensor Manipulation

" Reshaping - View - Stacking - Squeeze - Unsqueeze - premute " 

```python
import torch
x = torch.arange(1,10)
x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(3,3)          # Note that , the reshape size should be compatible to the no of exixting elements { 9 elements= 3x3 }
x_reshaped
```

```python
# change the view
z = x.view(3,3)
z                  # Changing the view z will change the x also ( because x and z use the same memory)
```

```python
# Stacking the tensors
x_stack = torch.stack([x,x,x] , dim=1)
x_stack
```

```python
# Squeeze removes the 1-dim elements from the data
x_reshaped = x_reshaped.squeeze().shape    # remove one sq. bracket == one dimensions
x_reshaped                                 # Unsqueese adds a dimension to the target tensor ( dim=1)
```

```python
# permute , changes the axis of dimension
x = torch.rand(224,224,3)
x_premute = x.permute(2,1,0)
x_premute.shape
```
